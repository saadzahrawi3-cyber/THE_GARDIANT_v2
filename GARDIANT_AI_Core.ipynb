```python
# 1. Installation des dépendances réseau et d'imagerie
!pip install flask pyngrok opencv-python numpy

import cv2
import numpy as np
from flask import Flask, request, jsonify
from pyngrok import ngrok

# 2. Authentification du tunnel Ngrok
# ⚠️ COLLEZ VOTRE TOKEN NGROK CI-DESSOUS ⚠️
ngrok.set_auth_token("3F3aTdT7lPYbR4Yi7yq2LmmCdXR_4gj57bGfK2uCSL1LDs2CA")

# 3. Initialisation du serveur Flask
app = Flask(__name__)

# 4. Création de la route d'API pour recevoir l'image de l'ESP32
@app.route('/api/upload', methods=['POST'])
def upload_frame():
    if not request.data:
        return jsonify({"status": "error", "message": "Aucun octet reçu"}), 400

    try:
        # Lecture des octets bruts (le fichier JPEG binaire) envoyés par l'ESP32
        img_bytes = request.data
        nparr = np.frombuffer(img_bytes, np.uint8)

        # Décodage avec OpenCV en une matrice d'image (Pixels RGB)
        img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

        if img is None:
            return jsonify({"status": "error", "message": "Fichier image corrompu"}), 400

        print(f"\n[SERVEUR CLOUD] 🟢 Succès ! Image réceptionnée.")
        print(f" -> Taille du fichier : {len(img_bytes)} octets")
        print(f" -> Résolution : {img.shape[1]}x{img.shape[0]} pixels")

        # ---------------------------------------------------------
        # SAUVEGARDE DE L'IMAGE POUR POUVOIR LA VISUALISER
        # ---------------------------------------------------------
        cv2.imwrite("derniere_capture.jpg", img)
        print(" -> 📸 Image sauvegardée ! Ouvrez le dossier à gauche pour la voir.")

        return jsonify({"status": "success", "message": "Trame sécurisée et sauvegardée"}), 200

    except Exception as e:
        return jsonify({"status": "error", "message": str(e)}), 500

# 5. Démarrage du tunnel sécurisé et affichage de l'URL publique
public_url = ngrok.connect(5000, bind_tls=False).public_url
print("\n" + "="*70)
print("🚀 SERVEUR EN LIGNE ! Copiez l'URL exacte ci-dessous dans votre code ESP32 :")
print(f"👉 {public_url}/api/upload")
print("="*70 + "\n")

# Démarrage du serveur web local dans Google Colab
app.run(port=5000, threaded=True)
```
